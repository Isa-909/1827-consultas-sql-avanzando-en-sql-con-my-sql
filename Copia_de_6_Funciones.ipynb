# Creación de Funciones
Así como hemos utilizado la función print(), range(), round() etc, en python podemos crear nuestras propias funciones. Existen dos tipos de funciones, parametrizadas y no parametrizadas.

Las funciones no parametrizadas no requieren recibir datos para poder operar, a diferencia de las parametrizadas que requieren recibir información.

Una función la definimos siempre al principio del código de la siguiente manera:

def NombreDeLaFunción():

      "Lo que deseamos que haga nuestra función"

Si la unción es parametrizada debemos poner el nombre de las variables de los parámetros

def NombreDeLaFunción(par1,par2,par3 ...):

      "Lo que deseamos que haga nuestra función"

Una vez que se definió la función, la podemos usar "invocandola" mediante su nombre: NombreDeLaFuncion(), y en ese momento se ejecutará su contenido.


```python
#Crear una fucnión de inicio del programa que indica el inicio dle programa
#Crea una función que imprime un mensaje de error
# función de fin de programa

#Apartado de funciones
def  inicio():
  print('Inicio del programa...')

def error():
  print('error')

def fin():
  print('fin del programa')

#Cuerpo del programa
inicio()
error()
fin()
```

##Funciones parametrizadas

Reciben información mediante variables, usan dicha infomación dentro de sus lineas de código, pueden o no devolver valores.

Las variables que se encuentren dentro de una función solo viven dentro de ella.



```python
#Crear una función que recibe un numero y lo devuelve en positivo
def absoluto(x):
  if x >= 0 :
    return x
  else:
    return -x

inicio()
num = float(input('dime un num '))
print(f'El absoluto de {num} es {absoluto(num)}')
fin()
```

```python
# Crear una función que calcula áreas de rectángulos de forma parametrizada
def area_rect_param(b,h):
  return b*h

inicio()
base = float(input('dame la base '))
altura = float(input('dame la altura '))
area = area_rect_param(base,altura)
print(f'El area es {area}')
fin()
```

```python
#Crear una función que reciba una lista de numeros, devuelve la lista en forma
#creciente y sin números repetidos
def nueva_lista(lista):
  for i in lista:
    n=2
    while n>1:
      n = lista.count(i)
      if n>1:
        lista.remove(i)
  return lista.sort()

lista = [1,2,3,4,1,2,1,2,4,5,6]
nueva_lista(lista)
lista
```

```python
'''Programa que identifica si un número es par o impar
a) Imprimiendo si se trata o no de número par
b) Devolviendo true si se trata de número par'''
def parimpar(n):
  if n%2 == 0:
    return True
  else:
    return False

numero = float(input('dame un num '))
if parimpar(numero):
  print(f'{numero} es par')
else:
  print(f'{numero} es impar')
```

1. Función para llenar una lista de números


```python
def llenar_lista_numeros():
    numeros = []
    while True:
        try:
            num = float(input("Ingrese un número (o 'fin' para terminar): "))
            numeros.append(num)
        except:
            break
    return numeros
```



2. Función que calcula la sumatoria



```python
def sumatoria(lista):
    total = 0
    for num in lista:
        total += num
    return total
```

3. Función promedio usando la función sumatoria


```python
def promedio(lista):
    if len(lista) == 0:
        return 0
    return sumatoria(lista) / len(lista)
```

4. Función para calcular perímetro de rectángulo

5. Función para calcular calificación final


```python
def calificacion_final(cal1, cal2, cal3):
    promedio = (cal1 + cal2 + cal3) / 3
    if promedio < 6:
        return promedio, "Te vas a extra"
    return promedio, "Aprobado"
```

8. Programa para calcular factoriales


```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n-1)

def calcular_factoriales():
    count = 0
    while True:
        try:
            num = int(input("Ingrese un número (o letra para terminar): "))
            print(f"Factorial de {num}: {factorial(num)}")
            count += 1
        except:
            break
    print(f"Total de números leídos: {count}")
```

9. Función para separar reprobados


```python
def reprobados(nombres, calificaciones):
    return [nombres[i] for i in range(len(nombres)) if calificaciones[i] < 6]
```

10. Funciones para manipular listas


```python
def ordena_creciente(lista):
    return sorted(lista)

def ordena_decreciente(lista):
    return sorted(lista, reverse=True)

def elimina_por_indice(lista, indice):
    return lista.pop(indice)

def elimina_por_dato(lista, dato):
    lista.remove(dato)
    return lista

def estadisticas(lista):
    if not lista:
        return 0, 0, 0
    return sum(lista)/len(lista), max(lista), min(lista)
```

11. Función para identificar números primos


```python
def es_primo(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

Ejemplo de uso con argumentos arbitrarios (*args)


```python
def suma_args(*args):
    return sum(args)

# Puede recibir cualquier cantidad de números
print(suma_args(1, 2, 3))       # 6
print(suma_args(5, 10, 15, 20))  # 50
```
