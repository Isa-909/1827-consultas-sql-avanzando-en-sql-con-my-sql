1. Programa que recibe 10 datos y los muestra elevados al cuadrado


```python
numeros = []
for i in range(10):
    num = float(input(f"Ingrese el número {i+1}: "))
    numeros.append(num)

print("\nNúmeros elevados al cuadrado:")
for num in numeros:
    print(f"{num}² = {num**2}")
```

2. Programa que recibe n calificaciones de n materias


```python
materias = []
calificaciones = []
n = int(input("¿Cuántas materias desea registrar? "))

for i in range(n):
    materia = input(f"Nombre de la materia {i+1}: ")
    calif = float(input(f"Calificación de {materia}: "))
    materias.append(materia)
    calificaciones.append(calif)

print("\nReporte de calificaciones:")
for i in range(len(materias)):
    print(f"{materias[i]}: {calificaciones[i]}")
```

4. Programa de inventario con opción mostrar todos


```python
productos = ["Laptop", "Mouse", "Teclado", "Monitor", "USB"]
claves = ["LP001", "MS002", "TC003", "MN004", "UB005"]
existencias = [15, 42, 28, 10, 35]

opcion = input("¿Desea buscar por índice (I) o ver todos (T)? ").upper()

if opcion == "I":
    indice = int(input("Ingrese el índice del producto a buscar (0-4): "))
    print(f"\nProducto: {productos[indice]}")
    print(f"Clave: {claves[indice]}")
    print(f"Existencias: {existencias[indice]}")
elif opcion == "T":
    print("\nInventario completo:")
    for i in range(len(productos)):
        print(f"{i}. {productos[i]} - Clave: {claves[i]} - Existencias: {existencias[i]}")
```

5. Programa de inventario con búsqueda múltiple


```python
productos = ["Laptop", "Mouse", "Teclado", "Monitor", "USB"]
claves = ["LP001", "MS002", "TC003", "MN004", "UB005"]
existencias = [15, 42, 28, 10, 35]

opcion = input("¿Buscar por índice (I), nombre (N) o clave (C)? ").upper()

if opcion == "I":
    indice = int(input("Ingrese el índice del producto a buscar (0-4): "))
    print(f"\nProducto: {productos[indice]}")
    print(f"Clave: {claves[indice]}")
    print(f"Existencias: {existencias[indice]}")
elif opcion == "N":
    nombre = input("Ingrese el nombre del producto: ")
    if nombre in productos:
        indice = productos.index(nombre)
        print(f"\nProducto: {productos[indice]}")
        print(f"Clave: {claves[indice]}")
        print(f"Existencias: {existencias[indice]}")
    else:
        print("Producto no encontrado")
elif opcion == "C":
    clave = input("Ingrese la clave del producto: ")
    if clave in claves:
        indice = claves.index(clave)
        print(f"\nProducto: {productos[indice]}")
        print(f"Clave: {claves[indice]}")
        print(f"Existencias: {existencias[indice]}")
    else:
        print("Clave no encontrada")
```

10. Programa avanzado de manipulación de listas


```python
tipo = input("¿Desea crear lista de números (N) o texto (T)? ").upper()
lista = []

if tipo == "N":
    while True:
        try:
            num = float(input("Ingrese un número (o 'fin' para terminar): "))
            lista.append(num)
        except:
            break
else:
    while True:
        texto = input("Ingrese texto (o 'fin' para terminar): ")
        if texto.lower() == 'fin':
            break
        lista.append(texto)

while True:
    print("\nOpciones:")
    print("1. Agregar valores")
    print("2. Eliminar valores (por índice)")
    print("3. Eliminar valores (por valor)")
    print("4. Ordenar lista (modificando)")
    print("5. Ordenar lista (creando nueva)")
    print("6. Buscar elemento")
    print("7. Operaciones numéricas (solo números)")
    print("8. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        if tipo == "N":
            try:
                num = float(input("Ingrese número a agregar: "))
                lista.append(num)
            except:
                print("Debe ingresar un número")
        else:
            texto = input("Ingrese texto a agregar: ")
            lista.append(texto)
        print("Lista actual:", lista)

    elif opcion == "2":
        try:
            indice = int(input("Ingrese índice a eliminar: "))
            if 0 <= indice < len(lista):
                eliminado = lista.pop(indice)
                print(f"Se eliminó: {eliminado}")
            else:
                print("Índice fuera de rango")
        except:
            print("Índice inválido")
        print("Lista actual:", lista)

    elif opcion == "3":
        valor = input("Ingrese valor a eliminar: ")
        if tipo == "N":
            try:
                valor = float(valor)
            except:
                print("Debe ingresar un número")
        if valor in lista:
            lista.remove(valor)
            print(f"Se eliminó: {valor}")
        else:
            print("Valor no encontrado")
        print("Lista actual:", lista)

    elif opcion == "4":
        lista.sort()
        print("Lista ordenada:", lista)

    elif opcion == "5":
        lista_ordenada = sorted(lista)
        print("Nueva lista ordenada:", lista_ordenada)
        print("Lista original:", lista)

    elif opcion == "6":
        buscar = input("Ingrese valor a buscar: ")
        if tipo == "N":
            try:
                buscar = float(buscar)
            except:
                print("Debe ingresar un número")
        if buscar in lista:
            indice = lista.index(buscar)
            print(f"El valor {buscar} está en el índice {indice}")
        else:
            print("Valor no encontrado")

    elif opcion == "7" and tipo == "N":
        print("\nOperaciones numéricas:")
        print(f"Máximo: {max(lista)}")
        print(f"Mínimo: {min(lista)}")
        print(f"Suma: {sum(lista)}")
        print(f"Promedio: {sum(lista)/len(lista)}")

    elif opcion == "8":
        print("Saliendo del programa...")
        break

    else:
        print("Opción inválida")
```

12. Programa de búsqueda sin usar 'in'


```python
lista = [2, 5, 8, 12, 16, 23, 38, 42]
valor = int(input("Ingrese valor a buscar: "))
encontrado = False
indice = -1

for i in range(len(lista)):
    if lista[i] == valor:
        encontrado = True
        indice = i
        break

if encontrado:
    print(f"El valor {valor} se encontró en el índice {indice}")
else:
    print(f"El valor {valor} no está en la lista")
```
