Textblob es una librería de procesamiento del texto para Python que permite realizar tareas de Procesamiento del Lenguaje Natural como análisis morfológico, extracción de entidades, análisis de opinión, traducción automática, etc. Está construida sobre otras dos librerías muy famosas de Python: NLTK y pattern.

```python
#Ingresa un texto en inglés
import textblob
from textblob import TextBlob
texto="It is very good, it is large and there is a lot of variety of products and the prices are good, they are worth it, there are some interesting offers, the attention of the workers is good and their willingness to help you find very good products there,"
t=TextBlob(texto)
print(t.sentiment)
```

#Función Input

a función input() es capaz de leer datos que fueron introducidos por el usuario y pasar esos datos al programa en ejecución.

El programa solicita al usuario que inserte algún dato desde la consola (seguramente utilizando el teclado, aunque también es posible introducir datos utilizando la voz o alguna imagen).

La función input() tiene la ventaja de poder funcionar como un print para poder dar instrucciones al usuario sobre el dato que se va a recibir. La función pondrá la consola en modo de entrada; aparecerá un cursor que parpadea, y podrás introducir datos con el teclado, al terminar presiona la tecla enter; todos los datos introducidos serán enviados al programa a través del resultado de la función.

Nota: el resultado debe ser asignado a una variable; esto es crucial, si no se hace los datos introducidos se perderán.

Nota: la función input siempre recibe los datos en formato de caracter, en caso de querer cambiarlo a numérico debemos usar la función int() o float()

7. Cálculo de beneficio

11. Cálculo para banquete

14. Interés simple y compuesto


```python
capital = float(input("Capital inicial: "))
tasa = float(input("Tasa de interés (%): ")) / 100
periodos = int(input("Periodos: "))
simple = capital * (1 + tasa * periodos)
compuesto = capital * (1 + tasa) ** periodos
print(f"Interés simple: ${simple:.2f}")
print(f"Interés compuesto: ${compuesto:.2f}")
```
