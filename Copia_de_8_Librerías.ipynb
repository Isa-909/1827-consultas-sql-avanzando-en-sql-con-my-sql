##Determinar el punto de equilibrio, precio de equilibrio y gráfica para:

Qd=500-25p

Qo=25p

```python
import sympy as sp
p,q=sp.symbols('p q')
```

```python
Qd=500-25*p
Qo=25*p
Ecu1=sp.Eq(Qd,Qo)
display("Oferta igual a demanda", Ecu1)
Sol1=sp.solve(Ecu1)
display("Precio de equilibrio", Sol1[0])
Qequi=Qd.subs(p,Sol1[0])
display("Cantidad de equilibrio",Qequi)
sp.plot(Qd,Qo, (p,0,2*Sol1[0]))
```

```python
#Simulador de 10 mercados distintos cambiando el parametro i
plista=[]
qlista=[]
for i in range (1,11):
  Qd=500*i-25*p
  Qo=25*p
  Ecu1=sp.Eq(Qd,Qo)
  #display("Oferta igual a demanda", Ecu1)
  Sol1=sp.solve(Ecu1)
  #display("Precio de equilibrio", Sol1[0])
  Qequi=Qd.subs(p,Sol1[0])
  #display("Cantidad de equilibrio",Qequi)
  plista.append(Sol1[0])
  qlista.append(Qequi)
display("Precios de equilibrio",plista,"Cantidades de equilibrio",qlista)
```

1. Operaciones con math


```python
import math as mt

# a) Seno de 45 + coseno de pi
a = mt.sin(mt.radians(45)) + mt.cos(mt.pi)
print(f"a) {a:.4f}")

# b) Raíz de euler elevado a la dos
b = mt.sqrt(mt.exp(2))
print(f"b) {b:.4f}")

# c) Redondear a 3 decimales la multiplicación entre pi y raiz de dos
c = round(mt.pi * mt.sqrt(2), 3)
print(f"c) {c}")
```

2. Resolución de ecuaciones con sympy

```python
import sympy as sp

x = sp.symbols('x')

# a) x^2 - 5x + 7 = 0
ecu_a = sp.Eq(x**2 - 5*x + 7, 0)
sol_a = sp.solve(ecu_a, x)
print("a) Soluciones:")
for i, sol in enumerate(sol_a):
    print(f"  x{i+1} = {sol}")

# b) 7x^2 + 9x -7 = 8x^2 - 2x - 3
ecu_b = sp.Eq(7*x**2 + 9*x - 7, 8*x**2 - 2*x - 3)
sol_b = sp.solve(ecu_b, x)
print("\nb) Soluciones:")
for i, sol in enumerate(sol_b):
    print(f"  x{i+1} = {sol}")
```

3. Creación de DataFrame con pandas


```python
import pandas as pd

# Crear el DataFrame
data = {
    'Producto': ['bimbo', 'alpura', 'suavitel', 'pachoncito'],
    'Precio': [45, 27, 72, 103],
    'Cantidad': [400, 350, 273, 321]
}

df = pd.DataFrame(data)

# Mostrar el DataFrame
print("DataFrame completo:")
print(df)

# Mostrar estadísticos principales
print("\nEstadísticos principales:")
print(df.describe())

# Estadísticos solo para precios
print("\nEstadísticos para precios:")
print(df['Precio'].describe())

# Estadísticos solo para cantidades
print("\nEstadísticos para cantidades:")
print(df['Cantidad'].describe())
```
