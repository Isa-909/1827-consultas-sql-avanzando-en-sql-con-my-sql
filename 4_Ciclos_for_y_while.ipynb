#CICLO FOR
El ciclo "for" es una estructura de control en programación que se utiliza para repetir un bloque de código un número específico de veces.

El ciclo "for" es especialmente útil cuando se conoce de antemano la cantidad de veces que se desea repetir un bloque de código. Es muy utilizado para recorrer elementos de una lista o un arreglo, realizar cálculos en secuencias numéricas, y en general, para realizar tareas que requieran un número fijo de iteraciones.

La sintaxis del ciclo es:

for i in range(n):
  #acción a repetir



```python
#Programa que muestra 10 holas mundo
# for VARIABLE in range(a):
#    AQUI LO QUE SE VA A REPETIR
# donde a es un numero entero positivo y representa la cantidad de veces a repetir

for i in range(10):
  print(f'{i+1}.- Hola mundo')
```

Ciclo While
1. Elevar al cuadrado hasta que el usuario decida parar

```python
while True:
    num = float(input("Ingresa un número para elevarlo al cuadrado: "))
    print(f"Resultado: {num**2}")
    continuar = input("¿Deseas continuar? (si/no): ").lower()
    if continuar != 'si':
        break
```

2. Interés compuesto continuo


```python
while True:
    K0 = float(input("Capital inicial: "))
    i = float(input("Tasa de interés (decimal): "))
    objetivo = float(input("Objetivo financiero: "))
    t = 0

    while K0 * (1+i)**t < objetivo:
        t += 1

    print(f"Se necesitan {t} periodos para alcanzar ${objetivo:.2f}")
    continuar = input("¿Otra simulación? (si/no): ").lower()
    if continuar != 'si':
        break
```

3. Validar número entre 1 y 5


```python
while True:
    num = int(input("Ingresa un número entre 1 y 5: "))
    if 1 <= num <= 5:
        print("Número válido!")
        break
    print("Número fuera de rango, intenta nuevamente.")
```

4. Continuar con variantes de "si"


```python
while True:
    respuesta = input("¿Deseas continuar? ")
    if respuesta.lower() not in ['si', 'sí']:
        break
```

5. Preguntar equipo favorito hasta que sea "América"


```python
while True:
    equipo = input("¿A qué equipo le vas? ").lower()
    if equipo == "américa":
        print("¡Correcto! Eres un verdadero fan.")
        break
    print("¡Equipo equivocado! Intenta nuevamente.")
```

6. Validar contraseña hasta coincidir


```python
while True:
    pass1 = input("Ingresa tu contraseña: ")
    pass2 = input("Confirma tu contraseña: ")
    if pass1 == pass2:
        print("Contraseña confirmada!")
        break
    print("Las contraseñas no coinciden, intenta nuevamente.")
```

7. Validar contraseña con 3 intentos


```python
intentos = 0
pass1 = input("Ingresa tu contraseña: ")

while intentos < 3:
    pass2 = input("Confirma tu contraseña: ")
    if pass1 == pass2:
        print("Contraseña confirmada!")
        break
    print("Las contraseñas no coinciden.")
    intentos += 1
else:
    print("Cuenta cancelada")
```

8. Sumar abonos hasta superar $100,000


```python
total = 0
while total <= 100000:
    abono = float(input("Cantidad a abonar: "))
    if abono < 0:
        print("No se permiten cantidades negativas")
        continue
    total += abono
    print(f"Total abonado: ${total:.2f}")
print("¡Meta alcanzada!")
```

9. Calculadora básica


```python
while True:
    print("\nOperaciones disponibles:")
    print("1. Suma\n2. Resta\n3. Multiplicación\n4. División\n5. Potencia\n6. Módulo")
    opcion = input("Elige una operación (1-6): ")

    a = float(input("Primer número: "))
    b = float(input("Segundo número: "))

    if opcion == '1':
        print(f"Resultado: {a + b}")
    elif opcion == '2':
        print(f"Resultado: {a - b}")
    elif opcion == '3':
        print(f"Resultado: {a * b}")
    elif opcion == '4':
        if b == 0:
            print("Error: división entre cero")
        else:
            print(f"Resultado: {a / b}")
    elif opcion == '5':
        print(f"Resultado: {a ** b}")
    elif opcion == '6':
        print(f"Resultado: {a % b}")
    else:
        print("Opción inválida")

    continuar = input("¿Deseas realizar otra operación? (si/no): ").lower()
    if continuar != 'si':
        break
```

10. Calculadora con repetición por operación


```python
while True:
    print("\nOperaciones disponibles:")
    print("1. Suma\n2. Resta\n3. Multiplicación\n4. División\n5. Potencia\n6. Módulo")
    opcion = input("Elige una operación (1-6): ")

    while True:
        a = float(input("Primer número: "))
        b = float(input("Segundo número: "))

        if opcion == '1':
            print(f"Resultado: {a + b}")
        elif opcion == '2':
            print(f"Resultado: {a - b}")
        elif opcion == '3':
            print(f"Resultado: {a * b}")
        elif opcion == '4':
            if b == 0:
                print("Error: división entre cero")
            else:
                print(f"Resultado: {a / b}")
        elif opcion == '5':
            print(f"Resultado: {a ** b}")
        elif opcion == '6':
            print(f"Resultado: {a % b}")
        else:
            print("Opción inválida")
            break

        repetir = input("¿Repetir esta operación? (si/no): ").lower()
        if repetir != 'si':
            break

    continuar = input("¿Deseas realizar otra operación diferente? (si/no): ").lower()
    if continuar != 'si':
        break
```
